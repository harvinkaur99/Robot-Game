```python
def greet_professor():
    """
    This function prints a polite greeting message
    at the start of the program.
    """
    print("Hello Prof. Dr. Iris Lorscheid, welcome to our Robot Game project!")
    print("This program demonstrates Python concepts such as functions and OOP.")
    print("-" * 60)

# Call the greeting function
greet_professor()


import random

class RobotGame:
    def __init__(self, rows=6, cols=7, energy=7, total_bricks=4):
        self.rows = rows
        self.cols = cols
        self.energy = energy
        self.steps = 0
        self.bricks_collected = 0
        self.total_bricks = total_bricks
        self.game_over = False

        # Create world (2D list)
        self.world = [["." for _ in range(cols)] for _ in range(rows)]

        # Robot start
        self.robot_pos = [0, 0]
        self.world[0][0] = "R"

         # Place home
        self.world[0][0] = "H"

        # Monster 1 start (last row, last column)
        self.monster1_pos = [rows - 1, cols - 1]
        self.world[self.monster1_pos[0]][self.monster1_pos[1]] = "M"

        # Monster 2 start (last row, first column)
        self.monster2_pos = [rows - 1, 0]
        self.world[self.monster2_pos[0]][self.monster2_pos[1]] = "M"

        # Place bricks
        self.place_bricks(total_bricks)

    def place_bricks(self, number):
        self.bricks = []
        while len(self.bricks) < number:
            r = random.randint(0, self.rows - 1)
            c = random.randint(0, self.cols - 1)
            if self.world[r][c] == ".":
                self.world[r][c] = "B"
                self.bricks.append((r, c))

    def print_world(self):
        for row in self.world:
            print(" ".join(row))
        print()
    # Defining robot movements
    def move_robot(self, direction):
        if self.energy <= 0 or self.game_over:
            return

        r, c = self.robot_pos
        new_r, new_c = r, c

        if direction == "up":
            new_r -= 1
        elif direction == "down":
            new_r += 1
        elif direction == "left":
            new_c -= 1
        elif direction == "right":
            new_c += 1
        else:
            print("Invalid direction.")
            return

        if not (0 <= new_r < self.rows and 0 <= new_c < self.cols):
            print("Robot hit the boundary.")
            return

        self.update_robot_position(new_r, new_c)
    # Moving the robot
    def update_robot_position(self, new_r, new_c):
        old_r, old_c = self.robot_pos
        if self.world[old_r][old_c] != "H":#Home should remain
            self.world[old_r][old_c] = "."

        self.robot_pos = [new_r, new_c]
        self.energy -= 1
        self.steps += 1
       # Collecting the Brick
        if self.world[new_r][new_c] == "B":
            self.bricks_collected += 1
            self.energy += 2
            print("Robot collected a brick!")
        # Colliding with the monster
        if self.world[new_r][new_c] == "M":
            self.game_over = True
            print("A monster caught the robot! Game over.")

        self.world[new_r][new_c] = "R"

        self.move_monsters()
        self.check_collision()
        self.print_world()
    # Moving the Monsters randomly
    def move_monsters(self):
        if self.game_over:
            return

        self.monster1_pos = self.move_single_monster(self.monster1_pos)
        self.monster2_pos = self.move_single_monster(self.monster2_pos)

    def move_single_monster(self, monster_pos):
        directions = ["up", "down", "left", "right"]
        direction = random.choice(directions)

        r, c = monster_pos
        new_r, new_c = r, c

        if direction == "up":
            new_r -= 1
        elif direction == "down":
            new_r += 1
        elif direction == "left":
            new_c -= 1
        elif direction == "right":
            new_c += 1

        if 0 <= new_r < self.rows and 0 <= new_c < self.cols:
            # Determine what to leave behind at the monster's old position
            if (r, c) in self.bricks:
                self.world[r][c] = "B" # Restore brick if monster was on it
            else:
                self.world[r][c] = "." # Otherwise, leave an empty spot

            # Update monster's position
            monster_pos = [new_r, new_c]

            # Check if monster moved onto the robot
            if self.world[new_r][new_c] == "R":
                self.game_over = True
                print("A monster moved onto the robot! Game over.")

            # Place monster at new position
            self.world[new_r][new_c] = "M"

        return monster_pos

    def check_collision(self):
        if self.robot_pos == self.monster1_pos or self.robot_pos == self.monster2_pos:
            self.game_over = True
            print("Robot collided with a monster! Game over.")
    # Printing the status of the robot
    def status(self):
        print("Steps:", self.steps)
        print("Energy:", self.energy)
        print("Bricks collected:", self.bricks_collected)
        print("Game over:", self.game_over)


```

```python
# -------- GAME LOOP --------

print("""Welcome to the Robot Game with TWO Monsters
         In order to win the game you should collect all the Bricks before your energy finishes while being safe from the Monsters.
         You should provide the robot with the moves in terms of directions.

                                                   """)
print("Allowed directions: up, down, left, right")
print("Separate moves with commas (example: right,down,left)")

game = RobotGame()
game.print_world()

while not game.game_over and game.energy > 0 and game.bricks_collected < game.total_bricks:
    print(f"Energy left: {game.energy}")
    user_input = input("Enter your moves: ").lower()
    moves = [m.strip() for m in user_input.split(",")]

    for move in moves:
        if game.game_over or game.energy <= 0 or game.bricks_collected == game.total_bricks:
            break
        game.move_robot(move)

game.status()

if game.bricks_collected == game.total_bricks:
    print("🎉 All bricks collected! You win!")
elif game.energy <= 0:
    print("⚡ Robot ran out of energy.")
elif game.game_over:
    print("☠️ The robot was defeated by the monsters.")
```
